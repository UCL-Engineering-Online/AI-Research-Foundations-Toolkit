## Overview



In this lab, you will explore how a model’s behavior can be shaped by the data it is trained on. Rather than focusing on a full machine learning framework, the activity uses a simple word-weight model to simulate the main ideas behind fine-tuning. You will create a small dataset of labeled input/output examples, define a basic prediction function, and test how the model behaves before any training has taken place. This helps show that a model does not automatically understand meaning, but instead relies on the patterns and parameters it has learned.

You will then run a simple training loop that adjusts word weights when the model makes incorrect predictions, allowing you to observe how training changes its behavior. After evaluating the model again, you will experiment with a biased dataset to investigate overfitting and poor generalization. By the end of the lab, you should understand how training data, repeated patterns and parameter updates influence model behavior, why biased or limited datasets can lead to unreliable predictions, and how better data can improve generalization to unseen inputs.
