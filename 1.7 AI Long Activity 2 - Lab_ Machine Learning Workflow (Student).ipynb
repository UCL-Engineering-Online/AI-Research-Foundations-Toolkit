```python
import numpy as np
import pandas as pd

# Upload the CSV file manually in VS Code or Jupyter.
UPLOAD_PATH = "co2 Emission Africa.csv"

df = pd.read_csv(UPLOAD_PATH)
df.head()
```

## Step 1: Inspect the dataset
- Look for missing values
- Identify numerical vs categorical variables
- Think about which features might predict CO₂ emissions

```python
df.info()
```

## Step 2: Clean the data
Fill or remove missing values. You may choose your own strategy.

```python
df_clean = df.fillna(df.mean(numeric_only=True))
df_clean.head()
```

## Step 3: Choose features and target
Pick numerical columns as predictors. Target = CO₂ emissions.

```python
# Example: choose numeric features automatically.
features = df_clean.select_dtypes(include=["float64", "int64"]).drop(columns=["Total CO2 Emission including LUCF (Mt)"])
target = df_clean["Total CO2 Emission including LUCF (Mt)"]

features.head()
```

## Step 4: Build a simple model
We will use a baseline linear regression model.

Feel free to try other models!

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(features, target)

preds = model.predict(features)
preds[:10]
```

## Step 5: Evaluate your model
Try different loss functions: MAE, MSE, etc.

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error

mae = mean_absolute_error(target, preds)
mse = mean_squared_error(target, preds)

mae, mse
```
