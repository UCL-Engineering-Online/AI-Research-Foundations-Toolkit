## Step 1 – Create a dataset

You define a small dataset of input/output pairs.

> 👉 **Interpretation:**  
> - This is your **fine-tuning dataset**.
> - It defines what the model will learn.
>
> **Key question (ask students):**  
> What behavior will the model learn from this dataset?


```python

data = [
    ("good service", 1),
    ("great support", 1),
    ("bad service", 0),
    ("terrible support", 0),
]

data
```


## Step 2 – Define a simple model

You simulate a model using **word weights**.

> 👉 **Idea:**
> - Positive words → positive score
> - Negative words → negative score

---

> 💭 **Reflection:**  
> Before running:
> - What will the model predict?


```python
# Initial weights.
# Starting from zero allows students to observe learning during training.
weights = {
    "good": 0,
    "great": 0,
    "bad": 0,
    "terrible": 0,
}


def predict(text):
    """Return 1 for positive prediction and 0 for negative prediction."""
    score = 0

    for word in text.split():
        score += weights.get(word, 0)

    if score >= 0:
        return 1

    return 0


predict("good service")
```


## Step 3 – Evaluate before training

You test the model on **new inputs**.

> 👉 This is like a **test set**

---

> 💭 **Reflection:**
> - Does the model generalize?
> - Where might it fail?


```python
test_data = [
    "nice service",
    "awful support",
    "great service",
    "bad support",
]

[(x, predict(x)) for x in test_data]
```


## Step 4 – Training loop

You simulate training:

- If prediction is wrong → update weights.
- This mimics **learning from data**.

---

### Key idea:
Training = adjusting parameters to reduce errors

---

### Reflection before running:
- What will change?


**Teacher note:** Initial weights should start at zero so that students observe parameters changing during training.

```python
# Training loop.
# Purpose:
# Adjust weights when predictions are incorrect.
for epoch in range(5):
    for text, label in data:
        pred = predict(text)

        if pred != label:
            for word in text.split():
                if label == 1:
                    weights[word] = weights.get(word, 0) + 1
                else:
                    weights[word] = weights.get(word, 0) - 1

weights
```


## Step 5 – Evaluate after training

Now you test again.

---

### Reflection:
- Did performance improve?
- Did behavior change?


```python
# Evaluate after training.
# Compare these outputs with Step 3.

[(x, predict(x)) for x in test_data]
```

## Step 6 – Overfitting experiment

You now create a dataset where one feature becomes misleading.

> ℹ️ **Info:**  
> The word "cheap" appears mostly in positive examples.
> This may cause the model to over-rely on that feature.

---

### Prediction:
- What will the model learn?
- Will it generalize correctly?


```python
# Overfitting-style dataset.
biased_data = [
    ("cheap phone", 1),
    ("cheap laptop", 1),
    ("cheap charger", 1),
    ("cheap tablet", 0),
]

# Reset weights.
weights = {}

# Train model.
for epoch in range(5):
    for text, label in biased_data:
        pred = predict(text)

        if pred != label:
            for word in text.split():
                if label == 1:
                    weights[word] = weights.get(word, 0) + 1
                else:
                    weights[word] = weights.get(word, 0) - 1

weights
```


## Evaluate biased model

> 💭 **Reflection:**  
> - What did the model learn?
> - Does it generalize?


```python
# Evaluate biased model.
test = [
    "cheap headphones",
    "expensive laptop",
    "cheap tablet",
]

[(x, predict(x)) for x in test]
```
