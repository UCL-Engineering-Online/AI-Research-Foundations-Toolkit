## 1. Define a small vocabulary

Start with the same tiny vocabulary as in the student notebook.  
You may adapt this to local context (for example, change the tokens to reflect school life, local transport, or local professions).


```python
tokens = [
    "school",
    "teacher",
    "student",
    "laptop",
    "phone",
    "market",
    "village",
    "city",
]

print("Tokens in our vocabulary:")

for t in tokens:
    print("-", t)
```

---
## 2. Create simple embeddings

Use **small random vectors** for simplicity.

> 📝 **Teacher notes:**  
> You can replace these by *hand-crafted* vectors if you want to encode simple intuitions, for example:
> - make `school`, `teacher`, `student` close together,
> - make `village` closer to `market` than to `laptop`, etc.
----
For now, stick to random vectors but keep the seed fixed so results are reproducible.


```python
import numpy as np

# Fix the random seed so that you get the same vectors every time.
np.random.seed(0)

d = 2  # Embedding dimension (2D is easier to reason about).
embeddings = {
    tok: np.random.randn(d)
    for tok in tokens
}

embeddings
```

You can print the vectors more clearly if you want to discuss them in class:

```python
for tok, vec in embeddings.items():
    print(
        f"{tok:8s} -> {vec}",
    )
```

---
## 3. Compute pairwise distances

Compute the **Euclidean distance** between all pairs of tokens.

> 📝 **Teacher notes:**  
> - Emphasize that distance is just a number measuring how far apart two vectors are.  
> - In higher dimensions, the intuition is the same even if you cannot draw the space.


```python
def euclidean_distance(u, v):
    """Return the Euclidean distance between two vectors u and v."""
    return float(np.linalg.norm(u - v))


n = len(tokens)
distance_matrix = np.zeros((n, n))

for i, ti in enumerate(tokens):
    for j, tj in enumerate(tokens):
        distance_matrix[i, j] = euclidean_distance(
            embeddings[ti],
            embeddings[tj],
        )

distance_matrix
```

For readability, you can also show the distance matrix as a labelled table:

```python
import pandas as pd

df_dist = pd.DataFrame(
    distance_matrix,
    index=tokens,
    columns=tokens,
)

df_dist
```

---
## 4. Nearest neighbours

For each token, identify the closest other token (smallest non-zero distance).  

> 📝 **Teacher notes:**  
> Ask students if these nearest neighbours match their intuition.  
> If not, it is a good moment to point out that **embeddings encode the dataset and training process**, not “truth”.


```python
for i, ti in enumerate(tokens):
    best_j = None
    best_d = float("inf")

    for j, tj in enumerate(tokens):
        if i == j:
            continue

        d = distance_matrix[i, j]

        if d < best_d:
            best_d = d
            best_j = j

    print(
        f"Nearest neighbour of '{ti}' is "
        f"'{tokens[best_j]}' "
        f"(distance {best_d:.3f})",
    )
```

## 5. Cosine similarity (Optional extension)

Cosine similarity measures the **angle** between two vectors, ignoring their length.  
It is often used in NLP because it is important to care about direction more than magnitude.

> 📝 **Teacher notes:**  
> You do not need to go into the full formula if the class is not comfortable with it.  
You can just emphasize:  
> “Cosine similarity is high when vectors point in similar directions.”


```python
def cosine_similarity(u, v):
    """Return the cosine similarity between two vectors."""
    num = float(np.dot(u, v))
    den = float(np.linalg.norm(u) * np.linalg.norm(v))

    if den == 0:
        return 0.0

    return num / den


print(
    "Cosine similarity between 'school' and 'student':",
    cosine_similarity(
        embeddings["school"],
        embeddings["student"],
    ),
)
print(
    "Cosine similarity between 'village' and 'city':",
    cosine_similarity(
        embeddings["village"],
        embeddings["city"],
    ),
)
```

You can also build a small cosine similarity table if you want students to compare:

```python
cos_matrix = np.zeros((n, n))

for i, ti in enumerate(tokens):
    for j, tj in enumerate(tokens):
        cos_matrix[i, j] = cosine_similarity(
            embeddings[ti],
            embeddings[tj],
        )

import pandas as pd

df_cos = pd.DataFrame(
    cos_matrix,
    index=tokens,
    columns=tokens,
)

df_cos
```

> 💭 **Reflection (for teachers):**
> **Key teaching points:**
> - Embeddings turn questions about meaning into questions about **geometry** (distances and angles).
> - Different initializations or training corpora will produce **different neighborhoods**.
> - If training data under-represents certain communities or languages, their tokens may end up with unstable or poor embeddings.
> - There is no single “correct” embedding space – it always reflects choices and data.
>
> **Possible discussion prompts:**
>
> - “Do the nearest neighbours make sense to you? If not, why might the model still behave this way?”  
> - “How could we design embeddings that better respect local languages or contexts?”  
> - “What might happen if most of the training data comes from one region or social group?”
