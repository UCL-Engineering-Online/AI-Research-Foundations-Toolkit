## Step 1 – Define the game

The function below simulates one game. You do not need to edit this function. Run the cell and read the docstring to understand the rules.


```
def play_game(deck, max_total=15):
    """Play one card-drawing game.

    Args:
        deck: A list of card values that can be drawn.
        max_total: The target total for the game.

    Returns:
        True if the game reaches exactly max_total.
        False if the game goes above max_total.
    """
    total = 0

    while total < max_total:
        card = random.choice(deck)
        total += card

        if total > max_total:
            return False

    return True
```

## Step 2 – Estimate a win rate

The function below plays many games and estimates the proportion of games that end in a win. You do not need to edit this function.


```
def win_rate(deck, n=5000):
    """Estimate the win rate for a deck using repeated simulation.

    Args:
        deck: A list of card values that can be drawn.
        n: Number of simulated games.

    Returns:
        The proportion of games that reached exactly 15.
    """
    wins = sum(play_game(deck) for _ in range(n))
    return wins / n
```

## Step 3 – Run the baseline experiment

The baseline deck contains the numbers 1–10, each appearing once. This means each card value is equally likely.


```
baseline_deck = list(range(1, 11))
baseline_rate = win_rate(baseline_deck)

baseline_rate
```
